Lab 1: Design and implement a basic simulation of a Federated Learning system in Python where multiple clients train local models on their own data without sharing raw datasets. A central server aggregates the locally trained model parameters using averaging to form a global model, demonstrating the core federated learning workflow.


```python
import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
```

```python
#create synthetic data

NUM_CLIENTS = 5
SAMPLES_PER_CLIENT = 200
FEATURES = 10

clients_data = []

for i in range(NUM_CLIENTS):
    X, y = make_classification(
        n_samples=SAMPLES_PER_CLIENT,
        n_features=FEATURES,
        n_informative=6,
        n_redundant=2,
        random_state=i
    )
    clients_data.append((X, y))
```

```python
# Define Local Client Training
# Each client: Receives the global model, Trains locally, Sends only model parameters back
# Raw data never leaves the client

def train_local_model(global_weights, X, y):
    model = LogisticRegression(max_iter=200)

    # Initialize model with global weights
    model.fit(X, y)

    return model.coef_, model.intercept_
```

```python
# Core: Federated Averaging (Server-Side)
def federated_averaging(local_weights):
    avg_coef = np.mean([w[0] for w in local_weights], axis=0)
    avg_intercept = np.mean([w[1] for w in local_weights], axis=0)
    return avg_coef, avg_intercept
```

```python
# Initialize Global Model
global_coef = np.zeros((1, FEATURES))
global_intercept = np.zeros(1)
```

```python
#Federated Learning Rounds
ROUNDS = 5

for round_num in range(ROUNDS):
    local_weights = []

    print(f"\n--- Federated Round {round_num + 1} ---")

    for client_id, (X, y) in enumerate(clients_data):
        coef, intercept = train_local_model(
            (global_coef, global_intercept), X, y
        )
        local_weights.append((coef, intercept))

    # Server aggregates models
    global_coef, global_intercept = federated_averaging(local_weights)
```

```python
#Evaluate Global Model

# Combine all data for evaluation
X_test = np.vstack([client[0] for client in clients_data])
y_test = np.hstack([client[1] for client in clients_data])

global_model = LogisticRegression()
global_model.coef_ = global_coef
global_model.intercept_ = global_intercept
global_model.classes_ = np.array([0, 1])

y_pred = global_model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)

print("\nGlobal Model Accuracy:", accuracy)
print("\n Derived from Local Weights:")
print("\nGlobal Model Coefficient:", global_coef)
print("\nGlobal Model Intercept:", global_intercept)
```

What's happening:

Clients train locally

No raw data sharing

Central server only aggregates weights

Uses Federated Averaging

Mimics real-world FL systems (Google Keyboard, healthcare ML)
